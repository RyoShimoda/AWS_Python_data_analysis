```python
# 平均注文額が最大だった月・州の行を、注文数とGMVも含めて確認する
max_aov_index = df_sales["average_order_value"].idxmax()

display(
    df_sales.loc[
        max_aov_index,
        [
            "purchase_month",
            "customer_state",
            "order_count",
            "gmv",
            "total_freight",
            "average_order_value",
        ],
    ]
)

# 全体の加重平均注文額を計算する
weighted_average_order_value = (
    df_sales["gmv"].sum() / df_sales["order_count"].sum()
)

print("全体の加重平均注文額:", round(weighted_average_order_value, 2))
```

```python
# 購入月別に集計
df_monthly = (
    df_sales
    .groupby("purchase_month", as_index=False)
    .agg(
        order_count=("order_count", "sum"),
        gmv=("gmv", "sum"),
        total_freight=("total_freight", "sum"),
    )
)

# 月別の加重平均注文額を計算する
df_monthly["average_order_value"] = (
    df_monthly["gmv"] / df_monthly["order_count"]
)

# 月の順に並べて表示する
df_monthly = df_monthly.sort_values("purchase_month").reset_index(drop=True)
display(df_monthly)
```

```python
import matplotlib.pyplot as plt

# 年月文字列を日付型に変換し、グラフの横軸に使う
df_monthly["purchase_month_date"] = pd.to_datetime(
    df_monthly["purchase_month"],
    format="%Y-%m"
)

# 上下に2つのグラフを作る
fig, axes = plt.subplots(2, 1, figsize=(8, 5), sharex=True)

# 月別GMVの推移
axes[0].plot(
    df_monthly["purchase_month_date"],
    df_monthly["gmv"],
    marker="o"
)
axes[0].set_title("Monthly GMV")
axes[0].set_ylabel("GMV")
axes[0].grid(True)

# 月別注文数の推移
axes[1].plot(
    df_monthly["purchase_month_date"],
    df_monthly["order_count"],
    marker="o",
    color="orange"
)
axes[1].set_title("Monthly Order Count")
axes[1].set_ylabel("Orders")
axes[1].set_xlabel("Purchase Month")
axes[1].grid(True)

plt.tight_layout()
plt.show()
```

```python
# 州別の売上・注文数を集計する
df_state = (
    df_sales
    .groupby("customer_state", as_index=False)
    .agg(
        order_count=("order_count", "sum"),
        gmv=("gmv", "sum"),
        total_freight=("total_freight", "sum"),
    )
)

# 州別の加重平均注文額を計算する
df_state["average_order_value"] = (
    df_state["gmv"] / df_state["order_count"]
)

# GMVの大きい順に並べる
df_state = df_state.sort_values(
    "gmv",
    ascending=False
).reset_index(drop=True)

display(df_state)
```

SP州が一番GMVが大きいが、全体の加重平均額137.04円よりは低い。

```python
# 全州に占める各州の注文数とGMVの割合を計算する
total_orders = df_state["order_count"].sum()
total_gmv = df_state["gmv"].sum()

df_state["order_share_pct"] = (
    df_state["order_count"] / total_orders * 100
)
df_state["gmv_share_pct"] = (
    df_state["gmv"] / total_gmv * 100
)

# GMVの大きい上位10州を表示する
display(
    df_state[
        [
            "customer_state",
            "order_count",
            "order_share_pct",
            "gmv",
            "gmv_share_pct",
            "average_order_value",
        ]
    ].head(10).round(2)
)
```

```python
# GMVの大きい上位3州を取り出す
top_3_states = df_state.head(3)

# 上位3州が全体に占める割合を合計する
top_3_order_share = top_3_states["order_share_pct"].sum()
top_3_gmv_share = top_3_states["gmv_share_pct"].sum()

print(f"上位3州の注文数シェア: {top_3_order_share:.2f}%")
print(f"上位3州のGMVシェア: {top_3_gmv_share:.2f}%")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 箱ひげ図: GMVの大きい順に州を並べる
state_order = (
    df_sales.groupby("customer_state", as_index=False)
    .agg(gmv=("gmv", "sum"))
    .sort_values("gmv", ascending=False)["customer_state"]
    .tolist()
)

top5_gmv_states = ["SP", "RJ", "MG", "RS", "PR"]
top5_avg_states = ["AP", "PB", "AC", "AL", "PA"]

# GMV上位5州（青グラデーション: 上位ほど濃い青）
color_map_gmv = dict(
    zip(top5_gmv_states, sns.color_palette("Blues_r", n_colors=len(top5_gmv_states)).as_hex())
)
# 平均注文額上位5州（オレンジグラデーション: 上位ほど濃いオレンジ）
color_map_avg = dict(
    zip(top5_avg_states, sns.color_palette("Oranges_r", n_colors=len(top5_avg_states)).as_hex())
)
color_map = {**color_map_gmv, **color_map_avg}

palette = [color_map.get(state, "#cacdd1") for state in state_order]

plt.figure(figsize=(8, 5))
sns.boxplot(
    data=df_sales, 
    x='customer_state', 
    y='gmv',
    order=state_order,
    palette=palette,
    hue="customer_state",
    hue_order=state_order
)
plt.title('GMV by Customer State')
plt.xlabel('Customer State')
plt.ylabel('GMV')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# 円グラフ: GMV上位5州および平均注文額上位5州を表示し、それ以外は others としてまとめる
top5_gmv_states = ["SP", "RJ", "MG", "RS", "PR"]
top5_avg_states = ["AP", "PB", "AC", "AL", "PA"]

selected_states = top5_gmv_states + top5_avg_states
other_states = [s for s in df_state['customer_state'] if s not in selected_states]

pie_labels = top5_gmv_states + top5_avg_states + ['others']
pie_values = [
    df_state.loc[df_state['customer_state'] == s, 'gmv_share_pct'].values[0]
    for s in (top5_gmv_states + top5_avg_states)
] + [df_state[df_state['customer_state'].isin(other_states)]['gmv_share_pct'].sum()]

# GMV上位5州（青グラデーション）
color_map_gmv = dict(
    zip(top5_gmv_states, sns.color_palette("Blues_r", n_colors=len(top5_gmv_states)).as_hex())
)
# 平均注文額上位5州（オレンジグラデーション）
color_map_avg = dict(
    zip(top5_avg_states, sns.color_palette("Oranges_r", n_colors=len(top5_avg_states)).as_hex())
)
color_map = {**color_map_gmv, **color_map_avg}
colors = [color_map.get(state, "#cacdd1") for state in pie_labels]

fig, ax = plt.subplots(figsize=(7, 6))
wedges, texts, autotexts = ax.pie(
    pie_values,
    labels=[s if v >= 3.0 else '' for s, v in zip(pie_labels, pie_values)],
    autopct=lambda p: f'{p:.1f}%' if p >= 3.0 else '',
    startangle=90,
    counterclock=False,
    wedgeprops={"linewidth": 0.5, "edgecolor": "white"},
    colors=colors,
    pctdistance=0.75,
    textprops={"fontsize": 10},
)

legend_labels = [f"{s} ({v:.1f}%)" if v >= 0.1 else f"{s} ({v:.2f}%)" for s, v in zip(pie_labels, pie_values)]
ax.legend(
    wedges,
    legend_labels,
    title="Customer State",
    loc="center left",
    bbox_to_anchor=(1, 0, 0.5, 1),
    fontsize=9,
)
ax.set_title("GMV Share Percentage\n"
    "(Top 5 GMV & Top 5 Average Order Value States)", fontsize=12)

if other_states:
    plt.figtext(
        0.5,
        0.01,
        f"others = {', '.join(other_states)}",
        ha='center',
        fontsize=8,
    )

plt.tight_layout()
plt.show()
```

```python
# 円グラフ: 注文数上位5州および平均注文額上位5州を表示し、
# それ以外は others としてまとめる
top5_order_states = ["SP", "RJ", "MG", "RS", "PR"]
top5_avg_states = ["AP", "PB", "AC", "AL", "PA"]

selected_states = top5_order_states + top5_avg_states
other_states = [
    s
    for s in df_state["customer_state"]
    if s not in selected_states
]

pie_labels = top5_order_states + top5_avg_states + ["others"]

pie_values = [
    df_state.loc[
        df_state["customer_state"] == s,
        "order_share_pct",
    ].values[0]
    for s in selected_states
] + [
    df_state.loc[
        df_state["customer_state"].isin(other_states),
        "order_share_pct",
    ].sum()
]

# 注文数上位5州（青グラデーション）
color_map_order = dict(
    zip(
        top5_order_states,
        sns.color_palette(
            "Blues_r",
            n_colors=len(top5_order_states),
        ).as_hex(),
    )
)

# 平均注文額上位5州（オレンジグラデーション）
color_map_avg = dict(
    zip(
        top5_avg_states,
        sns.color_palette(
            "Oranges_r",
            n_colors=len(top5_avg_states),
        ).as_hex(),
    )
)

color_map = {
    **color_map_order,
    **color_map_avg,
}

colors = [
    color_map.get(state, "#cacdd1")
    for state in pie_labels
]

fig, ax = plt.subplots(figsize=(7, 6))

wedges, texts, autotexts = ax.pie(
    pie_values,
    labels=[
        state if value >= 3.0 else ""
        for state, value in zip(pie_labels, pie_values)
    ],
    autopct=lambda percentage: (
        f"{percentage:.1f}%"
        if percentage >= 3.0
        else ""
    ),
    startangle=90,
    counterclock=False,
    wedgeprops={
        "linewidth": 0.5,
        "edgecolor": "white",
    },
    colors=colors,
    pctdistance=0.75,
    textprops={"fontsize": 10},
)

legend_labels = [
    (
        f"{state} ({value:.1f}%)"
        if value >= 0.1
        else f"{state} ({value:.2f}%)"
    )
    for state, value in zip(pie_labels, pie_values)
]

ax.legend(
    wedges,
    legend_labels,
    title="Customer State",
    loc="center left",
    bbox_to_anchor=(1, 0, 0.5, 1),
    fontsize=9,
)

ax.set_title(
    "Order Count Share Percentage\n"
    "(Top 5 Order Count & Top 5 Average Order Value States)",
    fontsize=12,
)

if other_states:
    plt.figtext(
        0.5,
        0.01,
        f"others = {', '.join(other_states)}",
        ha="center",
        fontsize=8,
    )

plt.tight_layout()
plt.show()
```

```python
# 州ごとの average_order_value の箱ひげ図
# 平均注文額が大きい順に並べる
state_order = (
    df_sales.groupby("customer_state", as_index=False)
    .agg(average_order_value=("average_order_value", "mean"))
    .sort_values("average_order_value", ascending=False)["customer_state"]
    .tolist()
)

top5_gmv_states = ["SP", "RJ", "MG", "RS", "PR"]
top5_avg_states = ["AP", "PB", "AC", "AL", "PA"]

# GMV上位5州（青グラデーション: 上位ほど濃い青）
color_map_gmv = dict(
    zip(top5_gmv_states, sns.color_palette("Blues_r", n_colors=len(top5_gmv_states)).as_hex())
)
# 平均注文額上位5州（オレンジグラデーション: 上位ほど濃いオレンジ）
color_map_avg = dict(
    zip(top5_avg_states, sns.color_palette("Oranges_r", n_colors=len(top5_avg_states)).as_hex())
)
color_map = {**color_map_gmv, **color_map_avg}

palette = [color_map.get(state, "#cacdd1") for state in state_order]

plt.figure(figsize=(8, 5))
sns.boxplot(
    data=df_sales,
    x="customer_state",
    y="average_order_value",
    order=state_order,
    hue="customer_state",
    hue_order=state_order,
    dodge=False,
    palette=palette,
    legend=False,
    showmeans=True,
    meanprops={
        "marker": "D",
        "markerfacecolor": "white",
        "markeredgecolor": "black",
        "markersize": 7,
    },
)
plt.title("Average Order Value by Customer State")
plt.xlabel("Customer State")
plt.ylabel("Average Order Value")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### 図の集計単位と使い方

州別GMVの箱ひげ図は「月×州のGMV」、平均注文額の箱ひげ図は「月×州の平均注文額」の分布を表す。箱の中央値は注文1件ごとの金額の中央値ではない。州の全注文に対する平均注文額を比べる場合は、上で求めた`df_state["average_order_value"]`（州のGMV合計÷州の注文数合計）を使う。注文額そのものの中央値には注文単位のデータが必要。
