```python
import matplotlib.pyplot as plt
import seaborn as sns

df_boxplot = df_order_state.assign(
    order_value=df_order_state["order_gmv"]
)

# 平均注文額が大きい順に並べる
state_average = (
    df_boxplot.groupby("customer_state")["order_value"]
    .mean()
    .sort_values(ascending=False)
)

state_order = state_average.index.tolist()

top5_gmv_states = ["SP", "RJ", "MG", "RS", "PR"]
top5_avg_states = state_order[:5]

# GMV上位5州（青グラデーション: 上位ほど濃い青）
color_map_gmv = dict(
    zip(
        top5_gmv_states,
        sns.color_palette(
            "Blues_r",
            n_colors=len(top5_gmv_states),
        ).as_hex(),
    )
)

# 平均注文額上位5州（オレンジグラデーション: 上位ほど濃いオレンジ）
color_map_avg = dict(
    zip(
        top5_avg_states,
        sns.color_palette(
            "Oranges_r",
            n_colors=len(top5_avg_states),
        ).as_hex(),
    )
)

# 平均注文額上位5州の色を優先する
color_map = {
    **color_map_gmv,
    **color_map_avg,
}

palette = [
    color_map.get(state, "#cacdd1")
    for state in state_order
]

plt.figure(figsize=(12, 7.5))

sns.boxplot(
    data=df_boxplot,
    x="customer_state",
    y="order_value",
    order=state_order,
    hue="customer_state",
    hue_order=state_order,
    palette=palette,
    dodge=False,
    legend=False,
    showfliers=False,
    showmeans=True,
    meanprops={
        "marker": "D",
        "markerfacecolor": "white",
        "markeredgecolor": "black",
        "markeredgewidth": 1.2,
        "markersize": 6,
    },
)

plt.title("Order GMV by Customer State — Sorted by Mean")
plt.xlabel("Customer State")
plt.ylabel("Order GMV (product prices, excluding freight)")
plt.xticks(rotation=45, ha="right")
plt.grid(axis="y", alpha=0.25)
plt.tight_layout()
plt.savefig("../images/Order_Value_Distribution_by_State_Sorted_by_Mean.png", dpi=160)
plt.show()
```

```python
from scipy.stats import kruskal

state_groups = [
    group["order_gmv"].dropna().to_numpy()
    for _, group in df_order_state.groupby("customer_state")
    if group["order_gmv"].notna().any()
]

h_statistic, p_value = kruskal(*state_groups)

print(f"Kruskal-Wallis H statistic: {h_statistic:.4f}")
print(f"p-value: {p_value:.6e}")
```

```python
import scikit_posthocs as sp

# 州間のDunn多重比較
dunn_result = sp.posthoc_dunn(
    df_order_state,
    val_col="order_gmv",
    group_col="customer_state",
    p_adjust="holm",
)

display(dunn_result.round(4))
```

```python
import numpy as np
import pandas as pd

alpha = 0.05
significant_pairs = []

states = dunn_result.columns.tolist()

# 重複を避けるため、行列の上三角部分だけを確認
for i, state_1 in enumerate(states):
    for state_2 in states[i + 1:]:
        adjusted_p_value = dunn_result.loc[state_1, state_2]

        if adjusted_p_value < alpha:
            significant_pairs.append(
                {
                    "state_1": state_1,
                    "state_2": state_2,
                    "adjusted_p_value": adjusted_p_value,
                }
            )

significant_pairs_df = (
    pd.DataFrame(significant_pairs)
    .sort_values("adjusted_p_value")
    .reset_index(drop=True)
)

display(significant_pairs_df)
```

```python
from scipy.stats import mannwhitneyu
import pandas as pd

# 州ごとの注文金額を事前に保持
state_values = {
    state: group["order_gmv"].dropna().to_numpy()
    for state, group in df_order_state.groupby("customer_state")
}

effect_size_results = []

for row in significant_pairs_df.itertuples(index=False):
    state_1 = row.state_1
    state_2 = row.state_2

    values_1 = state_values[state_1]
    values_2 = state_values[state_2]

    # Mann–WhitneyのU統計量
    u_statistic, _ = mannwhitneyu(
        values_1,
        values_2,
        alternative="two-sided",
        method="asymptotic",
    )

    # 順位相関係数（state_1を基準にした符号付き効果量）
    rank_biserial = (
        2 * u_statistic / (len(values_1) * len(values_2))
        - 1
    )

    if rank_biserial > 0:
        higher_state = state_1
    elif rank_biserial < 0:
        higher_state = state_2
    else:
        higher_state = "same"

    effect_size_results.append(
        {
            "state_1": state_1,
            "state_2": state_2,
            "higher_state": higher_state,
            "adjusted_p_value": row.adjusted_p_value,
            "rank_biserial": rank_biserial,
            "absolute_effect_size": abs(rank_biserial),
            "median_state_1": pd.Series(values_1).median(),
            "median_state_2": pd.Series(values_2).median(),
        }
    )

effect_size_df = (
    pd.DataFrame(effect_size_results)
    .sort_values(
        "absolute_effect_size",
        ascending=False,
    )
    .reset_index(drop=True)
)

display(effect_size_df)
```

```python
def classify_effect_size(value):
    value = abs(value)

    if value >= 0.5:
        return "large"
    elif value >= 0.3:
        return "medium"
    elif value >= 0.1:
        return "small"
    else:
        return "negligible"


effect_size_df["effect_size_level"] = (
    effect_size_df["absolute_effect_size"]
    .apply(classify_effect_size)
)

display(
    effect_size_df[
        [
            "state_1",
            "state_2",
            "higher_state",
            "adjusted_p_value",
            "rank_biserial",
            "absolute_effect_size",
            "effect_size_level",
            "median_state_1",
            "median_state_2",
        ]
    ]
)
```
