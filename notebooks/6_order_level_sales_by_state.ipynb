## 検定と効果量

p値と効果量の役割を分けて読む。


```python
# 27州の注文額分布を比較する。
h_statistic, p_value = kruskal_by_group(
    df_order_state, "customer_state", "order_gmv"
)
print(f"Kruskal-Wallis H statistic: {h_statistic:.4f}")
print(f"p-value: {p_value:.6e}")
```

```python
import scikit_posthocs as sp

# 州間のDunn多重比較
dunn_result = sp.posthoc_dunn(
    df_order_state,
    val_col="order_gmv",
    group_col="customer_state",
    p_adjust="holm",
)

display(dunn_result.round(4))
```

```python
# 対称なDunn行列の上三角だけを調べ、重複ペアを除く。
alpha = 0.05
significant_pairs_df = significant_dunn_pairs(dunn_result, alpha)
display(significant_pairs_df)
```

```python
# 有意な州ペアについて、効果量の向きと大きさを計算する。
effect_size_df = state_pair_effect_sizes(df_order_state, significant_pairs_df)
display(effect_size_df)
```

```python
# p値は差の確からしさ、効果量は差の大きさを示す。
effect_size_df["effect_size_level"] = (
    effect_size_df["absolute_effect_size"].apply(state_effect_size_label)
)
display(effect_size_df[[
    "state_1", "state_2", "higher_state", "adjusted_p_value",
    "rank_biserial", "absolute_effect_size", "effect_size_level",
    "median_state_1", "median_state_2",
]])
```
