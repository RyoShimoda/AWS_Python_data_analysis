## データの確認


```python
from pathlib import Path
import sys
PROJECT_ROOT = next(
    path for path in [Path.cwd(), *Path.cwd().parents]
    if (path / "src").is_dir() and (path / "notebooks").is_dir()
)
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

import pandas as pd
from src.project_paths import DATA_DIR
from src.analysis_plots import plot_review_boxplot
from src.review_stats import (
    kruskal_by_group, epsilon_squared, pairwise_effect_sizes, cliffs_delta_label,
)
df_delivery = pd.read_csv(DATA_DIR / "delivery_delay_review_summary.csv")
```

## 配送遅延日数の分布


```python
# 価格・送料と同じ箱ひげ図関数を使い、遅延日数のみ表示範囲を固定する。
fig = plot_review_boxplot(
    df_delivery, "delivery_delay_days", "Delivery Delay by Review Score",
    "Delivery Delay (days)", ylim=(-50, 50), reference_line=0,
)
```

## 検定と効果量


```python
stat_delivery_delay, p_delivery_delay = kruskal_by_group(
    df_delivery, "review_score", "delivery_delay_days"
)
print(f"[Delivery Delay Days] H={stat_delivery_delay:.4f}, p={p_delivery_delay:.3e}")
```

```python
import scikit_posthocs as sp

dunn_delivery = sp.posthoc_dunn(
    df_delivery,
    val_col="delivery_delay_days",
    group_col="review_score",
    p_adjust="bonferroni"
)

dunn_delivery
```

```python
median_summary_delivery = (
    df_delivery
    .groupby("review_score")[["delivery_delay_days"]]
    .median()
)

median_summary_delivery["delivery_delay_days_diff_vs_5"] = (
    median_summary_delivery["delivery_delay_days"]
    - median_summary_delivery.loc[5, "delivery_delay_days"]
)

median_summary_delivery
```

```python
mean_sum_delivery_delay_days = (
    df_delivery
    .groupby("review_score")[["delivery_delay_days"]]
    .mean()
)

mean_sum_delivery_delay_days["delivery_delay_days_diff_vs_5"] =(
    mean_sum_delivery_delay_days["delivery_delay_days"]
    - mean_sum_delivery_delay_days.loc[5, "delivery_delay_days"]
)

mean_sum_delivery_delay_days
```

```python
# 分布差の大きさをε²で確認する。
N = len(df_delivery)
K = df_delivery["review_score"].nunique()
epsilon_delivery = epsilon_squared(stat_delivery_delay, N, K)
print(f"Delivery Delay Days ε²: {epsilon_delivery:.6f}")
```

```python
# Dunn検定の調整済みp値と、各レビュー点数の効果量をまとめる。
effect_size_delivery = pairwise_effect_sizes(
    df_delivery, "delivery_delay_days", "review_score", dunn_delivery
).rename(columns={"group_1": "review_score_1", "group_2": "review_score_2"})
effect_size_delivery
```

```python
effect_size_delivery["effect_size"] = (
    effect_size_delivery["cliffs_delta"].apply(cliffs_delta_label)
)
effect_size_delivery.sort_values("abs_cliffs_delta", ascending=False)
```
