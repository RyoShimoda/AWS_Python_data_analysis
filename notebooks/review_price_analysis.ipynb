```python
import matplotlib.pyplot as plt
import seaborn as sns


def plot_review_boxplot(data, value_col, title, ylabel):
    review_order = sorted(data["review_score"].dropna().unique())

    review_palette = dict(
        zip(
            review_order,
            sns.color_palette("Blues", n_colors=len(review_order))
        )
    )

    # 各レビュー評価について、実際に描画されるひげの端を取得
    lower_whiskers = []
    upper_whiskers = []

    for score in review_order:
        values = (
            data.loc[data["review_score"] == score, value_col]
            .dropna()
        )

        q1 = values.quantile(0.25)
        q3 = values.quantile(0.75)
        iqr = q3 - q1

        lower_fence = q1 - 1.5 * iqr
        upper_fence = q3 + 1.5 * iqr

        # Tukey形式で実際にひげとなるデータ点
        lower_whiskers.append(
            values.loc[values >= lower_fence].min()
        )
        upper_whiskers.append(
            values.loc[values <= upper_fence].max()
        )

    visible_min = min(lower_whiskers)
    visible_max = max(upper_whiskers)

    # 上下に余白を追加し、ひげの先端を軸から離す
    visible_range = visible_max - visible_min
    padding = visible_range * 0.08

    y_min = visible_min - padding
    y_max = visible_max + padding

    fig, ax = plt.subplots(figsize=(5, 3))

    sns.boxplot(
        data=data,
        x="review_score",
        y=value_col,
        order=review_order,
        hue="review_score",
        hue_order=review_order,
        palette=review_palette,
        dodge=False,
        legend=False,
        showfliers=False,
        showmeans=True,
        width=0.7,
        meanprops={
            "marker": "D",
            "markerfacecolor": "white",
            "markeredgecolor": "black",
            "markeredgewidth": 1.2,
            "markersize": 6,
        },
        ax=ax
    )

    ax.set_ylim(y_min, y_max)
    ax.set_title(title, fontsize=15, pad=12)
    ax.set_xlabel("Review Score", fontsize=12)
    ax.set_ylabel(ylabel, fontsize=12)
    ax.grid(axis="y", alpha=0.25)
    ax.set_axisbelow(True)

    plt.tight_layout()
    plt.show()
```

```python
plot_review_boxplot(
    data=df_review_order,
    value_col="total_price",
    title="Order Price by Review Score",
    ylabel="Total Price"
)
```

```python
from scipy.stats import kruskal

groups_price = [
    group["total_price"].values
    for _, group in df_review_order.groupby("review_score")
]

stat_price, p_price = kruskal(*groups_price)

print("=== Kruskal-Wallis Test ===")
print()
print("[Total Price]")
print(f"H-statistic: {stat_price:.4f}")
print(f"p-value: {p_price:.6g}")
```

```python
# total_freightの分布がレビュー点数ごとに異なるかどうかを検定
groups_freight = [
    group["total_freight"].values
    for _, group in df_review_order.groupby("review_score")
]

stat_freight, p_freight = kruskal(*groups_freight)

print()
print("[Total Freight]")
print(f"H-statistic: {stat_freight:.4f}")
print(f"p-value: {p_freight:.6g}")
```

```python
# Kruskal-Wallis統計量とグループ数 (k)、全サンプル数 (n) を基に、効果量ε²を計算

n = len(df_review_order)
k = df_review_order["review_score"].nunique()

epsilon_price = (stat_price - k + 1) / (n - k)
epsilon_freight = (stat_freight - k + 1) / (n - k)

print(f"Total Price  ε²: {epsilon_price:.6f}")
print(f"Total Freight ε²: {epsilon_freight:.6f}")
```

```python
import numpy as np
import pandas as pd
from itertools import combinations
from scipy.stats import mannwhitneyu


def calculate_pairwise_effect_size(
    data,
    value_col,
    group_col,
    dunn_result
):
    """各グループの組み合わせについてCliff's deltaを算出する。"""

    results = []

    groups = sorted(data[group_col].dropna().unique())

    for group_1, group_2 in combinations(groups, 2):
        x = data.loc[
            data[group_col] == group_1,
            value_col
        ].dropna().to_numpy()

        y = data.loc[
            data[group_col] == group_2,
            value_col
        ].dropna().to_numpy()

        # Uは「xがyより大きい確率」を基にした統計量
        u_stat = mannwhitneyu(
            x,
            y,
            alternative="two-sided",
            method="asymptotic"
        ).statistic

        cliffs_delta = (
            2 * u_stat / (len(x) * len(y)) - 1
        )

        results.append({
            "group_1": group_1,
            "group_2": group_2,
            "n_1": len(x),
            "n_2": len(y),
            "median_1": np.median(x),
            "median_2": np.median(y),
            "median_diff_1_minus_2": (
                np.median(x) - np.median(y)
            ),
            "dunn_p_adjusted": dunn_result.loc[
                group_1, group_2
            ],
            "cliffs_delta": cliffs_delta,
            "abs_cliffs_delta": abs(cliffs_delta)
        })

    return pd.DataFrame(results)
```

```python
def interpret_cliffs_delta(delta):
    abs_delta = abs(delta)

    if abs_delta < 0.147:
        return "negligible"
    elif abs_delta < 0.330:
        return "small"
    elif abs_delta < 0.474:
        return "medium"
    else:
        return "large"
```

```python
effect_size_freight = calculate_pairwise_effect_size(
    data=df_review_order,
    value_col="total_freight",
    group_col="review_score",
    dunn_result=dunn_freight
)

effect_size_freight["effect_size"] = (
    effect_size_freight["cliffs_delta"]
    .apply(interpret_cliffs_delta)
)

effect_size_freight
```
